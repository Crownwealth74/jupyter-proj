## DATA TYPE AND DATA STRUCTURE

```python
# Integer
x = 10
print("Integer:", x)

# Float
y = 3.14
print("Float:", y)

# Complex
z = 2 + 3j
print("Complex:", z)

# String
name = "Olamide"
print("String:", name)

# Boolean
is_active = True
print("Boolean:", is_active)
```

## DATA STRUCTURES

```python
# List (ordered, allows duplicates)
my_list = [1, 2, 3, 4, 5]
print("List:", my_list)

# Tuple (ordered, immutable)
my_tuple = (10, 20, 30)
print("Tuple:", my_tuple)

# Set (unordered, no duplicates)
my_set = {1, 2, 3, 3, 4}
print("Set:", my_set)

# Dictionary (key-value pairs)
my_dict = {"name": "Olamide", "age": 23, "course": "Information Technology"}
print("Dictionary:", my_dict)
```
